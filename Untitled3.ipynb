```python
# =====================================================================
# 1. IMPORTAR AS BIBLIOTECAS E FERRAMENTAS
# =====================================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# =====================================================================
# 2. CRIAR E GARANTIR A TABELA DE DADOS (df) NA MEMÓRIA
# =====================================================================
idades = [23, 45, 31, 22, 54, 38, 25, 41, 20, 48]
salarios = [22000, 45000, 35000, 19000, 60000, 28000, 26000, 52000, 18000, 40000]
scores = [450, 710, 580, 410, 780, 510, 490, 690, 390, 620]
incumprimentos = [1, 0, 0, 1, 0, 1, 0, 0, 1, 0]

dados_banco = {
    'idade': idades,
    'salario_anual': salarios,
    'score_credito': scores,
    'incumprimento': incumprimentos
}

df = pd.DataFrame(dados_banco)

# =====================================================================
# 3. GERAR OS GRÁFICOS VISUAIS
# =====================================================================
sns.set_theme(style="whitegrid")
plt.figure(figsize=(15, 5))

# GRÁFICO 1: Distribuição do Score de Crédito por Risco
plt.subplot(1, 2, 1)
sns.boxplot(data=df, x='incumprimento', y='score_credito', palette='Set2')
plt.title('Distribuição do Score de Crédito vs Incumprimento', fontsize=12, pad=15)
plt.xlabel('Histórico de Incumprimento (0 = Pagou | 1 = Falhou)', fontsize=10)
plt.ylabel('Score de Crédito do Cliente', fontsize=10)

# GRÁFICO 2: Cruzamento de Salário vs Score de Crédito
plt.subplot(1, 2, 2)
sns.scatterplot(data=df, x='salario_anual', y='score_credito', hue='incumprimento',
                style='incumprimento', palette='coolwarm', s=120, alpha=0.9)
plt.title('Zona de Risco: Salário Anual vs Score de Crédito', fontsize=12, pad=15)
plt.xlabel('Salário Anual (€)', fontsize=10)
plt.ylabel('Score de Crédito', fontsize=10)
plt.legend(title='Incumprimento', labels=['Pagou (0)', 'Falhou (1)'])

plt.tight_layout()
plt.show()
```
